```python
import pandas as pd;
import matplotlib.pyplot as plt
df=pd.read_csv("/content/funnel_events_sample.csv")
print(df.head())
print(df.shape)
print(df.info())

```

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()
```

```python
stage_users = (
      df.groupby('step')['user_id']
      .nunique()
      .reset_index(name='Users')
)

print(stage_users)
```

```python
order = [
    'visited_site',
    'signup_started',
    'details_filled',
    'email_verified',
    'purchase_completed'
]

stage_users['step'] = pd.Categorical(
    stage_users['step'],
    categories=order,
    ordered=True
)

stage_users = stage_users.sort_values('step')
```

Conversion rate calculation

```python
stage_users['Conversion_Rate'] = (
    stage_users['Users']
    / stage_users['Users'].shift(1)
    * 100
)

stage_users.loc[0, 'Conversion_Rate'] = 100
```

```python
display(stage_users)
```

Drop of percentage

```python
stage_users['drop_off_percentage'] = (
    100-stage_users['Conversion_Rate']
)
```

```python
display(stage_users)
```

```python
stage_users['Users_Lost'] = (
    stage_users['Users'].shift(1) - stage_users['Users']
)

stage_users['Users_Lost'] = stage_users['Users_Lost'].fillna(0).astype(int)
```

Largest drop

```python
largest_drop = stage_users.iloc[1:].loc[
    stage_users.iloc[1:]['drop_off_percentage'].idxmax()
]

print(largest_drop)
```

time calculation

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

```python
df = df.sort_values(['user_id', 'timestamp'])
```

```python
df['next_step'] = df.groupby('user_id')['step'].shift(-1)
df['next_timestamp'] = df.groupby('user_id')['timestamp'].shift(-1)
```

```python
df['time_to_next_stage'] = (
    df['next_timestamp'] - df['timestamp']
).dt.total_seconds() / 60
```

```python
valid_transitions = {
    ('visited_site', 'signup_started'),
    ('signup_started', 'details_filled'),
    ('details_filled', 'email_verified'),
    ('email_verified', 'purchase_completed')
}

if 'time_to_next_stage' not in df.columns:
    df['time_to_next_stage'] = (
        df['next_timestamp'] - df['timestamp']
    ).dt.total_seconds() / 60

df_filtered = df[
    df.apply(lambda x: (x['step'], x['next_step']) in valid_transitions, axis=1)
]
average_time = (
    df_filtered.groupby(['step', 'next_step'])['time_to_next_stage']
    .mean()
    .reset_index()
)
```

```python
print("\nAverage Time to Next Stage")
print(average_time)
```
